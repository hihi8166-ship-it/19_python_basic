```python
#for 와 range 사용하기
#for 변수 in range():
#    반복한 코드
# for 반복문 : 정해진 횟수만큼 반복
# while 반복문 : 조건이 참인 동안 반복

for i in range(100):
    print("hello, world?")
        
```

```python
#반복문에서 변수의 변화 알아보기

for i in range(10):
    print("hello, world", i)
```

```python
#시작하는 숫자와 끝나는 숫자 지정하기
for a in range(5, 12): #5부터 11까지 반복
    print("hello", a)
```

```python
#증가폭
for b in range(0, 10, 2):
    print("hello", b)
```

```python
#숫자 감소시키기

for i in range(10, 0): #range(10, 0)은 동작하지 않음
    print("hello", i)
```

```python
for i in range(10, 0, -1):
    print("hello", i)
```
