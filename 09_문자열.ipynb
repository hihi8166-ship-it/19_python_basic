```python
s = "python isn't difficult" #작은 따옴표를 감싸기위해 큰따옴표로 감쌌다. 작은따옴표 사용하면 에러나옴
s
```

```python
s='he said "python is easy"' 
print(s)
# 큰따옴표를 감싸기 위해 작은따옴표 사용, 만약 큰따옴표, 작은따옴표 모두 문장에 있을때는 """로 감싸면됨 또는 문장 사이에 \ 사용하면 됨
```

```python
s='python isn\'t difficult"' #'가 여러군데 나올때 '앞에 \입력하면 됨
print(s)
```

```python
s = """'python' is a "programming language"\n that lets you work quickly\n and\n integrate systems more eggectively.""" 
print(s)
```
