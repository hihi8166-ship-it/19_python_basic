```python
for i in range(1,101):
    if i % 3 == 0:
        print("Fizz") #15배수(3과 5의 공배수)의 경우 fizz 출력됨
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1,101):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz") #15배수(3과 5의 공배수)의 경우  
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
#코드 단축하기
for i in range(1, 101):
    print("Fizz" * (i % 3 == 0) + "Buzz" * ( i % 5 == 0) or i) #True = 1, False = 0
```

```python
for i in range(1, 101):
    print("Fizz" * (i % 3 == 0) and "Buzz" * ( i % 5 == 0) or i) #True = 1, False = 0 #and는 두가지 조건을 모두 만족하면 뒤에꺼 실행
```

```python
#20.7 연습문제
for i in range(1, 101):
    if i % 2 == 0 and i % 11 == 0:
        print("FizzBuzz")
    elif i % 2 == 0:
        print("Fizz")
    elif i % 11 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
#20.8 심사문제
a, b = map(int, input().split())
if a>=b:
    print("잘못된 범위입니다")
for x in range(a, b+1):
    if x % 35 == 0:
        print("FizzBuzz")
    elif x % 5 == 0:
        print("Fizz")
    elif x % 7 == 0:
        print("Buzz")
    else:
        print(x) #x를 i로 잘못 써버렸음. 이런 실수가 계속 나옴
```
