```python
#중첩루프 사용하기
for i in range(5):
    for j in range(5):
        print("j:", j, sep="", end=" ")
    print("i", i, "\\n", sep="")
    
```

```python
# 사각형 으로 별 출력하기 (p.214)
for i in range(5):
    for j in range(5):
        print("*", end="") #i가 0일때 j값 0~4까지 실행하고나서 
    print() #줄바꾸기
        
```

```python
# 사각형 모양 바꾸기
for i in range(3):
    for j in range(7):
        print("*", end="")
    print()
```

```python
#계단식으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j <= i:
            print("*", end="") #i=0, j=0이고 다시 위로, i=0, j=1 만족안하므로 밑에 프린트 실행
    print()
```

```python
#대각선으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
        else:
            print("*", end="")
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
#19.5 연습문제
for i in range(5):
    for j in range(5):
        if j < i:
            print(" ", end="")
        else:
            print("*", end="")
    print()
```

```python
#19.6 심사문제
n = int(input())
for i in range(n):
    print("_" * (n-i-1) + "*" * (2*i+1)) #별출력 공식
    
    
    
```
