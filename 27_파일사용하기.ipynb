```python
# 외부에 있는 파일을 가지고 오는 방법
#파일객체 = open(파일이름, 파일모드)
#파일객체.write(문자열)
#파일객체.close()

file = open("hello.txt", "w")  #./ 와 ""은 현재위치
file.write("hello world")
file.close()
```

```python
#파일에서 문자열 읽기

file = open("hello.txt", "r") #hello.txt 파일을 읽기모드(r)로 열기. 파일 객체 반환
s = file.read()               # 파일에서 문자열 읽기
print(s)                      # hello, world!
file.close()                  # 파일객체 닫기
```

```python
#자동으로 파일객체 닫기
with open("hello2.txt", "w") as file: #hello.txt 파일을 읽기모드(r) , 쓰기모드(w) 로 열기
    s = file.read()                  #파일에서 문자열 읽기
    print(s)                         #hello, world!
```

```python
with open("hello2.txt", "w") as file: #hello.txt 파일을 읽기모드(r) , 쓰기모드(w) 로 열기 (with 쓰면 close 안씀)
    for i in range(3):
        file.write(f"hello, world {i}\n")
```
