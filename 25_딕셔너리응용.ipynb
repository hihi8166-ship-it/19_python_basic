```python
#딕셔너리에 키와 기본값 저장하기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
x
```

```python
x.setdefault("e") #딕셔너리에 요소 추가
x
```

```python
x.setdefault("f", 100)
x
```

```python
#딕셔너리에 키의 값 수정하기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
x.update(a=90)
x
```

```python
x["a"] = 100 #이렇게 해도 됨
x
```

```python
#딕셔너리에 키와 값 쌍을 삭제하기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
x.pop("a") #매소드 이기 때문에 ()로 입력, 키값을 넣으면 삭제 (값 넣으면 에러뜸)
x
```

```python
x.clear()
x
```

```python
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
del x['a']
x
```

```python
#임의의 키값 쌍 삭제하기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
x.popitem()
```

```python
x
```

```python
#딕셔너리에서 키의 값 가져오기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
x.get("a")
```

```python
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
x.get("z")
```

```python
#딕셔너리에서 키값 쌍을 모두 가져오기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
x.items() #키와 밸류의 모든 값을 가져올때 딕트 형태로 나옴
```

```python
x.keys() #키값 가져옴
```

```python
x.values() #밸류값 가져옴
```

```python
#반복문으로 딕셔너리의 키값 쌍으로 모두 출력하기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
for i in x:
    print(i, end=" ")
```

```python
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
for i in x.items():
    print(i)
```

```python
for key, value in x.items():
    print(key, value)
```
