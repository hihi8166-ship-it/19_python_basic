```python
# 함수의 결과를 반환하기
# def 함수이름(매개변수):
#     return 반환값

def add(a, b):
    return a + b
x = add(10, 20)
print(x)
```

```python
# 함수에서 값을 여러 개 반환하기
def add_sub(a, b):
    return a+b, a-b

result = add_sub(10, 20)
print(result)
```

```python
#29.7 연습문제 (/ : 나누기(10/3=3.333), // : 몫 구하기(10//3=3), $ : 나머지 구하기(10%3=1)
x = 10
y = 3

def get_quotient_remainder(a, b):
    return a // b, a % b

quotient, remainder = get_quotient_remainder(x, y)
print("몫: {0}, 나머지: {1}".format(quotient, remainder))
```

```python
#29.8 심화문제  (f-string 방식 => 변수 이름으로 값을 넣는다.)
x, y = map(int, input().split())

def calc(x, y):
    return x+y, x-y, x*y, x/y

a, s, m, d = calc(x, y)
print(f"덧셈: {a}, 뺄셈: {s}, 곱셈: {m}, 나눗셈: {d}") #f는 문자열 안의 {}를 실제 값으로 바꿔주는 역할
```

```python
#29.8 심화문제 (.format 방식 => 순서번호로 값을 넣는다)
x, y = map(int, input().split())

def calc(x, y):
    return x+y, x-y, x*y, x/y

a, s, m, d = calc(x, y)
print("덧셈: {0}, 뺄셈: {1}, 곱셈: {2}, 나눗셈: {3}".format(a, s, m, d))
```
