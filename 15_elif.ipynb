```python
#if 조건식:
#    수행할코드
#elif 조건식:
#    수행할코드
    
x = 20
if x == 10:
    print("10입니다")
elif x == 20:
    print("20입니다")
```

```python

x = 30
if x == 10:
    print("10입니다")
elif x == 20:
    print("20입니다")
else:
    print("10도 20도 아닙니다") #else는 마지막에 나와야지 중간에 입력하면 오류뜸
```

```python
# 음료수 자판기 만들기
button = int(input())

if button == 1:
    print("콜라")
elif button == 2: #if 써도 값은 같은데 실행을 한번 돌리는거랑, if가 3개면 값을 3번 돌리게 된다.결국 실행횟수 차이
    print("사이다")
elif button == 3:
    print("환타")
else:
    print("제공하지 않는 메뉴")
```

```python
# 15.3 연습문제
x = int(input())

if 11 <= x <= 20:
    print("11~20")
elif 21 <= x <= 30:
    print("21~30")
else:
    print("아무것도 해당하지 않음")
```

```python
#15.4 심사문제 (처음에는 9000을 각각 모두 입력했는데, 잔액을 지정하고 잔액 balance 을 입력하는 방법으로 수정)
age = int(input())
balance = 9000

if 7 <= age <= 12:
    print(balance - 650)
elif 13 <= age <= 18:
    print(balance - 1050)
elif 17 <= age:
    print(balance - 1250)
else:
    print(balance)
```

```python
# 강사님 답안
age = int(input())
balance = 9000

if 7 <= age <= 12:
    balance = balance - 650
elif 13 <= age <= 18:
    balance = balance - 1050
elif 19 <= age:
    balance = balance - 1250
print(balance)
```
