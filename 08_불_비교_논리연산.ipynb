```python
#불과 비교 연산자 사용하기
print(True)
print(type(True))
print(False)
print(type(False))
```

```python
#비교연산자의 판단결과
3 > 1
print(3 > 1)
print(1 > 3)
```

```python
#숫자가 같은지 다른지 비교 == 비교연산자 같다. != 다르다. >= 크거나 같다. <= 작거나 같다.
print(10 == 10)
print(10 != 10)
print(10 > 10)
print(10 < 10)
print(10 >= 10)
print(10 <= 10)
```

```python
#논리연산자 사용하기
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

```python
# or 논리연산자
print(True or True)
print(True or False)
print(False or True)
print(False or False)
```

```python
#not x
print(not True)
print(not False)
```

```python
#(not True) and False or (not False)
not True and False or not False
```

```python
#논리연산자와 비교연산자를 함께 사용하기
print(10 == 10 and 10 != 5)
print(10 > 5 or 10 < 3)
print(not 10 > 3)
1 is 10.0
not(1 is 10.0)
#print(id(1))
#print(id(1.0))
```
