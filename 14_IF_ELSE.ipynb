```python
# 14.7 심사문제
국어, 영어, 수학, 과학 = map(int, input().split())
x = (국어+영어+수학+과학)/4
if x >= 80:
    print("합격")
    print("평균 점수는", x, "입니다.") #평균점수 노출방법
else:
    print("불합격") 
```

```python
국어, 영어, 수학, 과학 = map(int, input().split())
avg = (국어+영어+수학+과학)/4
if 국어 < 0 or 국어 > 100 or 영어 < 0 or 영어 > 100 or 수학 < 0 or 수학 > 100 or 과학 < 0 or 과학 > 100:
    print("잘못된 점수")
elif avg >= 80:
    print("합격")
    print("평균 점수는", x, "입니다.") #평균점수 노출방법
else:
    print("불합격") 
```
