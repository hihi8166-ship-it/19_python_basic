```python
for i in s:
    print(f"{i:8}") #8칸 확보
```

```python
for i in s:
    print(f"{i:0.2f}") #소수점 둘째자리
```

```python
for i in s:
    print(f"{i:10.2f}") #소수점 둘째자리 + 10칸 확보
```
