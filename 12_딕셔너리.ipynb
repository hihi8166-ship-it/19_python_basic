```python
#딕셔너리 만들기
#딕셔너리 = {키1:값1, 키2:값2, ...}
lux = {
    "health":490, #문자앞뒤로 " 사용, dict()안에서는 "사용안함
    "mana":334,
    "melee":550,
    "armor":18.72
}
lux
```

```python
# 키 이름이 중복이 되면 앞에서 인식못하고 뒤에꺼만 최신본으로 인식하여 출력함
lux = {
    "health":490,
    "health":334,
    "melee":550,
    "armor":18.72
}
lux
```

```python
# 딕셔너리의 키의 자료형
x = {
    100:"hundred",
    False:0,
    3.5:[3.5,3.5]
}
x
```

```python
# 딕셔너리의 키에 접근하고, 값 할당하기
lux={'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
print(lux["health"])
print(lux["armor"])
```

```python
# 딕셔너리의 키에 값 할당하기, 딕셔너리에서 값을 꺼낼때는 []로 꺼낸다.
lux={'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
lux["health"] = 2037
lux["mana"] = 1184
lux
```

```python
# 딕셔너리에서는 추가가 된다. 밑에 보면 원래 추가안된다.
lux["mana_regen"]=3.28
print(lux)
```
