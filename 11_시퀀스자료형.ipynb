```python
#del로 요소 삭제하기
#del 시퀀스객체[인덱스]
# del 시퀀스 객체는 하나씩 입력해야됨. 2개 동시에 삭제안됨( del a[2,3] 이런거 안됨. 한줄씩 써야됨)
# 삭제값을 다시 취소하고 원래값으로 돌아가는 거는 인설트 명령어가 있는데 다음시간에 나온다고 함
a = [38, 21, 53, 62, 19]
del a[2] 
a.insert(2, 53) #그럼 2번째 뒤에 삽입하는 방법은?
a
```
