```python
#반복문으로 2차원 리스트의 요소를 모두 출력하기
#for 반복문을 한번만 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for x, y in a:
    print(x, y)
```

```python
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for i in a:
    print(i)
```

```python
#for 반복문을 두 번 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for i in a:
    for j in i:
        print(j, end=" ")
    print() #줄바꿈 하겠다.
```

```python
#for와 range 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for i in range(len(a)): #0,1,2
    for j in range(len(a[i])): # 2니까 0,1 두개 사용됨
        print(a[i][j], end=" ")
    print()
```

```python
#whilt 반복문을 한 번 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```
