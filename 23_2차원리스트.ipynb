```python
#반복문으로 2차원 리스트의 요소를 모두 출력하기
#for 반복문을 한번만 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for x, y in a:
    print(x, y)
```

```python
#for 반복문을 두 번 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for i in a:
    for j in i:
        print(j, end=" ")
    print() #줄바꿈 하겠다.
```

```python
#for와 range 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for i in range(len(a)): #0,1,2
    for j in range(len(a[i])): # 2니까 0,1 두개 사용됨
        print(a[i][j], end=" ")
    print()
```

```python
#whilt 반복문을 한 번 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
#23.7 심사문제 지뢰찾기

row, col = map(int, input().split())
print(row)

matrix = []
for _ in range(row):
    matrix.append(list(input()))
    
   
```

```python
matrix
```

```python
for i in range(row):
    for j in range(col):
        if matrix[i][j] == "*":
            print("*", end="")
        else:
            print("ㅁ", end="")
        print()
```

```python
for i in range(row):
    for j in range(col):
        if matrix[i][j] == "*":
            print("*", end="")
        else:
            count = 0
            for r in range(i - 1, i + 2):
                for c in range(j - 1, j + 2):
                    if 0 <= r < row and 0 <= c < col:
                        if matrix[r][c] == "*":
                            count += 1
            print(count, end="")
    print() #위치에 따라 답이 달라짐
```
