```python
#위치 인수
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)
    
```

```python
print_numbers(10, 20, 30)
```

```python
#언패킹 사용하기
x = [10, 20, 30]
print_numbers(x) #하나의 객체라서 하나로 인식
```

```python
x = [10, 20, 30]
print_numbers(*x) # * 를 이용하여 언패킹 하는 방법
```

```python
x, y, z = [10, 20, 30]
print(x)
print(y)
print(z)
```

```python
# 가변인수 함수 만들기
# def 함수이름(*매개변수):
#     코드

def print_numbers(*args):
    print(args)
    for arg in args:
        print(arg)
```

```python
print_numbers(10)
```

```python
print_numbers(10, 20)
```

```python
def print_numbers(*args):
    # print(args)
    for arg in args:
        print(arg)
```

```python
print_numbers(10, 20, 30)
```

```python
# 키워드 인수 사용하기

def personal_info(name, age, address):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
personal_info("홍길동", 30, "서울시 용산구")
```

```python
personal_info(age=30, address="서울시", name="홍길동") # 순서 상관없이 입력해도 됨
```

```python
#키워드 인수와 딕셔너리 언패킹 사용하기 ????
    x = {
        "name": "홍길동",
        "age": 30,
        "address": "서울시 용산구"
    }
    
    personal_info(**x)
    personal_info(age=30, address="서울시", name="홍길동")
```

```python
# 가변 키워드 인수 함수 만들기
# def 함수이름(**매개변수):
#     코드

def personal_info(**kwargs):
    # print(kwargs)
    for kw, arg in kwargs.items():
        print(kw, arg)
```

```python
personal_info(name="홍길동", age=30)
```

```python
#매개변수에 초깃값 지정하기

def personal_info(name, age, address="비공개"):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
personal_info("홍길동", 30)
```

```python
personal_info("홍길동", 30, "서울시 용산구")
```

```python
def personal_info(name, address="비공개", age): #디폴트값 설정 다음에는 모두 디폴트로 설정해야됨
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
def personal_info(name, address="비공개", age=0): #디폴트값 설정 다음에는 모두 디폴트값을 설정해야됨
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```
