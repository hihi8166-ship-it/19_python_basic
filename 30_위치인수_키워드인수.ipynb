```python
#위치 인수
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)
    
```

```python
print_numbers(10, 20, 30)
```

```python
#언패킹 사용하기
x = [10, 20, 30]
print_numbers(x) #하나의 객체라서 하나로 인식

# def test(*args): 는 여러개의 값을 받아 하나의 튜플로 묶음
# test(*x) 는 리스트를 풀어서 여러 개의 값으로 전달
```

```python
x = [10, 20, 30]
print_numbers(*x) # * 를 이용하여 언패킹 하는 방법
```

```python
x, y, z = [10, 20, 30]
print(x)
print(y)
print(z)
```

```python
# 가변인수 함수 만들기 (함수에 숫자를 몇개 넣을지 정해져 있지 않을때 *args 사용 
# def 함수이름(*매개변수):
#     코드

def print_numbers(*args):         # def print_numbers : 함수 만들기 / *args : 전달받은 값들을 튜플로 모으기
    print(args)                   # 모인값 전체 출력
    for arg in args:              # 모인값을 하나씩 꺼내기
        print(arg)                # 꺼낸값을 출력
```

```python
print_numbers(10)
```

```python
print_numbers(10, 20)
```

```python
def print_numbers(*args):
    # print(args)
    for arg in args:
        print(arg)
```

```python
print_numbers(10, 20, 30)
```

```python
# 키워드 인수 사용하기

def personal_info(name, age, address):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
personal_info("홍길동", 30, "서울시 용산구")     # 1. 순서대로 전달
```

```python
personal_info(age=30, address="서울시", name="홍길동") # 2. 이름을 지정해서 전달 (키워드 인수 했으므로 순서 상관없이 입력해도 됨)
```

```python
#키워드 인수와 딕셔너리 언패킹 사용하기  (*x : 리스트나 튜플 등의 값을 하나씩 풀어서 전달, **x : 딕셔너리의 키와 값을 이름이 지정된 인수로 전달)
x = {
        "name": "홍길동",
        "age": 30,
        "address": "서울시 용산구"
    }
    
personal_info(**x)                                     # 딕셔너리를 풀어서 함수에 전달하라
    # personal_info(age=30, address="서울시", name="홍길동")
```

```python
# 가변 키워드 인수 함수 만들기
# def 함수이름(**매개변수):
#     코드

def personal_info(**kwargs):            # **kwargs : 이름과 값을 딕셔너리로 묶음 (별표가 2개임), 저장형태를 튜플이 아닌 딕셔너리임. 
    # print(kwargs)
    for kw, arg in kwargs.items():      # kwargs.items() 는 딕셔너리의 키와 값을 한 쌍씩 꺼내는 역할
        print(kw, arg)
```

```python
personal_info(name="홍길동", age=30)
```

```python
#매개변수에 초깃값 지정하기, 기본값(디폴트값), 주소를 입력하지 않으면 자동으로 "비공개"를 사용하라는 뜻.(주소입력하면 입력값 나옴)

def personal_info(name, age, address="비공개"):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
personal_info("홍길동", 30)
```

```python
personal_info("홍길동", 30, "서울시 용산구")
```

```python
def personal_info(name, address="비공개", age): #디폴트값 설정 다음에는 모두 디폴트로 설정해야됨
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
def personal_info(name, address="비공개", age=0): #디폴트값 설정 다음에는 모두 디폴트값을 설정해야됨
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```
